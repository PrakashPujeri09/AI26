```python
import random
def heuristic(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def display_board(state):
    n = len(state)

    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# N-Queens using Hill Climbing
def n_queens():
    n = int(input("Enter N: "))

    if n < 1:
        print("Invalid input")
        return

    if n == 2 or n == 3:
        print("No solution exists for N =", n)
        return

    state = [random.randrange(n) for _ in range(n)]
    h = heuristic(state)
    step = 1

    print("\nInitial state:", state)
    print("Initial h(n):", h)

    while h > 0:
        print("\nStep", step)

        best_state = None
        best_h = float("inf")

        for col in range(n):
            for row in range(n):
                if row != state[col]:
                    neighbor = state.copy()
                    neighbor[col] = row

                    neighbor_h = heuristic(neighbor)

                    print("State:", neighbor,
                          "h(n):", neighbor_h)

                    if neighbor_h < best_h:
                        best_h = neighbor_h
                        best_state = neighbor

        if best_h < h:
            state = best_state
            h = best_h
            print("Selected state:", state)
            print("New h(n):", h)
        else:
            print("No improvement. Restarting...")
            state = [random.randrange(n) for _ in range(n)]
            h = heuristic(state)
            print("New random state:", state)
            print("New h(n):", h)

        step += 1

    print("\nFinal chessboard:")
    display_board(state)

    print("\nSolution Found!")
    print("Number of attacking pairs:", h)


n_queens()
```
