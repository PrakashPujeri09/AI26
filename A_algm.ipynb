```python

import heapq

def heuristic(state, goal):
    h = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            h += 1

    return h


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(new_state)

    return neighbors


# A* Algorithm
def a_star(start, goal):

    pq = []

    # Initial g, h and f
    g = 0
    h = heuristic(start, goal)
    f = g + h

    heapq.heappush(pq, (f, g, start, []))

    visited = set()
    step = 0

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        step += 1

        print("\nStep:", step)

        print("State:")
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])

        print("g(n) =", g)
        print("h(n) =", heuristic(state, goal))
        print("f(n) =", f)

        # Check goal
        if state == goal:
            print("\nGoal reached!")
            print("Number of moves:", g)
            return

        for neighbor in get_neighbors(state):

            if tuple(neighbor) not in visited:

                new_g = g + 1
                new_h = heuristic(neighbor, goal)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, neighbor, path + [state])
                )


# Goal state
goal = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]

# Initial state
start = [
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
]

# Run A*
a_star(start, goal)

```

```python
def manhattan_distance(state, goal):
    h = 0

    for i in range(9):
        if state[i] != 0:

            # Current position
            row = i // 3
            col = i % 3

            # Goal position
            goal_index = goal.index(state[i])
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            # Manhattan distance
            h += abs(row - goal_row) + abs(col - goal_col)

    return h


# Initial state
state = [
    2, 8, 3,
    1, 6, 4,
    0, 7, 5
]

# Goal state
goal = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]

print("Manhattan Distance =", manhattan_distance(state, goal))
```
