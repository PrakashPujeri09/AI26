```python
import random
import math

def calculate_cost(board):
    cost = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                cost += 1
    return cost

def print_board(board):
    for row in range(8):
        for col in range(8):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def simulated_annealing():
    alpha = 0.995
    max_attempts = 100

    for attempt in range(1, max_attempts + 1):
        board = [random.randint(0, 7) for _ in range(8)]
        current_cost = calculate_cost(board)
        T = 100
        step = 1

        print("\nAttempt:", attempt)
        print("Initial Board:")
        print_board(board)
        print("Initial Cost:", current_cost)

        while T > 0.1 and current_cost != 0:
            col = random.randint(0, 7)
            new_row = random.randint(0, 7)

            while new_row == board[col]:
                new_row = random.randint(0, 7)

            new_board = board.copy()
            new_board[col] = new_row
            next_cost = calculate_cost(new_board)
            delta = current_cost - next_cost

            print("\nStep:", step)
            print("Temperature:", round(T, 2))
            print("Proposed Cost:", next_cost)

            if delta > 0:
                board = new_board
                current_cost = next_cost
                print("Better move accepted")
            elif random.random() < math.exp(delta / T):
                board = new_board
                current_cost = next_cost
                print("Worse move accepted")
            else:
                print("Move rejected")

            print("Current Cost:", current_cost)
            print("Current Board:")
            print_board(board)

            T *= alpha
            step += 1

        if current_cost == 0:
            print("\nSolution Found!")
            print("Final Board:")
            print_board(board)
            print("Final Cost:", current_cost)
            return

    print("\nSolution Not Found after", max_attempts, "attempts.")

simulated_annealing()
```
