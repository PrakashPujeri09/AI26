```python

import random

board = [" "] * 9

def show():
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])

def win(x):
    for a, b, c in [(0,1,2),(3,4,5),(6,7,8),
                    (0,3,6),(1,4,7),(2,5,8),
                    (0,4,8),(2,4,6)]:
        if board[a] == board[b] == board[c] == x:
            return True
    return False

while True:
    show()

    # Human
    p = int(input("Enter position (1-9): ")) - 1

    if board[p] != " ":
        print("Already taken!")
        continue

    board[p] = "X"

    if win("X"):
        show()
        print("You win!")
        break

    if " " not in board:
        show()
        print("Draw!")
        break

    # Computer
    empty = [i for i in range(9) if board[i] == " "]
    p = random.choice(empty)
    board[p] = "O"
    print("Computer played:", p + 1)

    if win("O"):
        show()
        print("Computer wins!")
        break


```
